```python
import pandas as pd

# Step 1: Load dataset
# Adding 'on_bad_lines="skip"' to ignore malformed lines that cause ParserError
df = pd.read_csv("styles.csv", on_bad_lines="skip")
print("Dataset:", df.shape)
print(df.head())

# Step 2: Select 100 records
df = df.head(100)

# Step 3: Create target
df["Target"] = df["masterCategory"].apply(
    lambda x: "Yes" if x == "Apparel" else "No"
)

features = ["gender", "subCategory", "articleType",
            "baseColour", "season", "usage"]

# Step 4: Find-S
h = ["ϕ"] * len(features)
print("\nInitial Hypothesis:", h)

for _, row in df.iterrows():
    if row["Target"] == "Yes":
        for i, f in enumerate(features):
            if h[i] == "ϕ":
                h[i] = row[f]
            elif h[i] != row[f]:
                h[i] = "?"

        print("Updated:", h)

# Step 5: Final hypothesis
print("\nFinal Hypothesis:", h)
```
