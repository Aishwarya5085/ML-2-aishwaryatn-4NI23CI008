```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from google.colab import files

# ------------------------------------------------------------
# 2. Upload the dataset
# ------------------------------------------------------------
print("Please upload Mall_Customers.csv")

uploaded = files.upload()

# Get the uploaded filename automatically
filename = list(uploaded.keys())[0]

# ------------------------------------------------------------
# 3. Read the dataset
# ------------------------------------------------------------
df = pd.read_csv(filename)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
print(df.head())

# ------------------------------------------------------------
# 4. Check dataset information
# ------------------------------------------------------------
print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 5. Select features for clustering
# ------------------------------------------------------------
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]

print("\nFeatures used for clustering:")
print(X.head())

# ------------------------------------------------------------
# 6. Visualize original data
# ------------------------------------------------------------
plt.figure(figsize=(8, 6))

plt.scatter(
    X['Annual Income (k$)'],
    X['Spending Score (1-100)'],
    color='blue',
    s=50
)

plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.title('Original Customer Data')

plt.show()

# ------------------------------------------------------------
# 7. Elbow Method
# ------------------------------------------------------------
inertia = []

for k in range(1, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X)

    inertia.append(kmeans.inertia_)

# Plot Elbow Curve
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    inertia,
    marker='o',
    color='blue'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Finding Optimal K')

plt.xticks(range(1, 11))
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# 8. Apply K-Means with 5 clusters
# ------------------------------------------------------------
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

# Assign each customer to a cluster
df['Cluster'] = kmeans.fit_predict(X)

# ------------------------------------------------------------
# 9. Display clustered data
# ------------------------------------------------------------
print("\n========================================")
print("CLUSTERED DATA")
print("========================================")

print(df.head(10))

# ------------------------------------------------------------
# 10. Count customers in each cluster
# ------------------------------------------------------------
print("\n========================================")
print("NUMBER OF CUSTOMERS IN EACH CLUSTER")
print("========================================")

print(df['Cluster'].value_counts().sort_index())

# ------------------------------------------------------------
# 11. Get cluster centers
# ------------------------------------------------------------
centers = kmeans.cluster_centers_

print("\n========================================")
print("CLUSTER CENTERS")
print("========================================")

print(centers)

# ------------------------------------------------------------
# 12. Visualize K-Means clusters
# ------------------------------------------------------------
plt.figure(figsize=(10, 7))

plt.scatter(
    X['Annual Income (k$)'],
    X['Spending Score (1-100)'],
    c=df['Cluster'],
    cmap='viridis',
    s=100,
    alpha=0.8
)

# Plot centroids
plt.scatter(
    centers[:, 0],
    centers[:, 1],
    c='red',
    marker='X',
    s=300,
    label='Centroids'
)

plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.title('Customer Segmentation using K-Means')
plt.colorbar(label='Cluster')
plt.legend()

plt.show()

# ------------------------------------------------------------
# 13. Cluster analysis
# ------------------------------------------------------------
print("\n========================================")
print("CLUSTER ANALYSIS")
print("========================================")

cluster_summary = df.groupby('Cluster')[
    ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
].mean()

print(cluster_summary.round(2))

# ------------------------------------------------------------
# 14. Save clustered dataset
# ------------------------------------------------------------
df.to_csv('Mall_Customers_Clustered.csv', index=False)

print("\n========================================")
print("PROJECT COMPLETED SUCCESSFULLY!")
print("========================================")

print("\nThe clustered dataset has been saved as:")
```
